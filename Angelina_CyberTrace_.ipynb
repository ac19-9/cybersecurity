## Section 4 — Suspicious Login Detector
4 row-level rules to flag suspicious login records.

```python
# ============================================================
# SECTION 4 — SUSPICIOUS LOGIN DETECTOR
# ============================================================

def get_suspicious_reason(login):
    """
    4 rules — returns a short badge label string, or None.

    Rule 1 — Repeated failures  : 3+ failed attempts on one record
    Rule 2 — After hours        : login hour between 00:00 and 04:59
    Rule 3 — Bad location       : Unknown or Tor Network
    Rule 4 — Flagged account    : phantom_x or unknown_scan
    """
    # Rule 1 — Repeated failed attempts
    if login["status"] == "Failed" and login["attempts"] >= 3:
        return f"⚠️  {login['attempts']} failed attempts"

    # Rule 2 — After-hours login (midnight to 4:59 AM)
    hour = int(login["timestamp"].split(" ")[1].split(":")[0])
    if 0 <= hour < 5:
        return f"🌙 After hours ({hour:02d}:xx)"

    # Rule 3 — Suspicious location
    if login["location"] == "Unknown" or "Tor" in login["location"]:
        return f"🌍 Location: {login['location']}"

    # Rule 4 — Flagged account name
    if login["username"] in ("phantom_x", "unknown_scan"):
        return f"👻 Flagged account: {login['username']}"

    return None  # Not suspicious by any rule


def render_logins():
    """Prints the login table with suspicious flags."""
    print(f"\n{'='*110}")
    print("  LOGIN ATTEMPTS TABLE")
    print(f"{'='*110}")
    header = f"{'ID':>3}  {'Timestamp':<20}  {'Username':<16}  {'IP':<18}  {'Location':<26}  {'Status':<8}  {'Att':>3}  Flag"
    print(header)
    print("-"*110)
    for login in login_attempts:
        reason = get_suspicious_reason(login)
        flag   = reason if reason else "✔  Normal"
        print(f"{login['id']:>3}  {login['timestamp']:<20}  {login['username']:<16}  "
              f"{login['ip']:<18}  {login['location']:<26}  {login['status']:<8}  "
              f"{login['attempts']:>3}  {flag}")

    susp_count = sum(1 for l in login_attempts if is_suspicious(l))
    normal_count = len(login_attempts) - susp_count
    print(f"{'='*110}")
    print(f"  Quick Summary: Scanned {len(login_attempts)} records. "
          f"🔴 {susp_count} suspicious  |  ✅ {normal_count} normal")
    print(f"  → See Section 5 (Alerts Panel) for deeper pattern analysis.")
    print(f"{'='*110}\n")

# NOTE: called in Section 10
```

2. Multiple usernames from one IP (credential stuffing)  
3. Successful login after repeated failures (brute-force win)

```python
# ============================================================
# SECTION 5 — ALERTS PANEL
# ============================================================

def detect_repeated_failures():
    """
    Detection 1: Repeated Failed Logins
    Any record where status='Failed' AND attempts >= 5.
    Threshold is higher than the row rule to surface only
    the most serious cases here.
    """
    alerts = []
    for login in login_attempts:
        if login["status"] == "Failed" and login["attempts"] >= 5:
            severity = "Critical" if login["attempts"] >= 9 else "High"
            alerts.append({
                "type":        "Repeated Failed Logins",
                "severity":    severity,
                "icon":        "🔁",
                "username":    login["username"],
                "ip":          login["ip"],
                "location":    login["location"],
                "timestamp":   login["timestamp"],
                "detail":      f"{login['attempts']} failed login attempts for account "
                               f"'{login['username']}' from IP {login['ip']} ({login['location']}).",
                "explanation": "A brute-force attack tries many passwords rapidly until one works. "
                               f"{login['attempts']} failures in a short window is a strong indicator. "
                               "Real analysts would lock the account after 5-10 failures and alert the security team."
            })
    return alerts


def detect_multi_user_same_ip():
    """
    Detection 2: Multiple Usernames from One IP
    Any IP that appears with 2 or more different usernames.
    Indicates credential stuffing / username enumeration.

    Steps:
      1. Group logins by IP into a dict: { ip -> set of usernames }
      2. For each IP, if unique username count >= 2, raise an alert.
    """
    alerts = []

    # Step 1 — build mapping: ip -> { usernames, locations, timestamps }
    ip_map = {}
    for login in login_attempts:
        ip = login["ip"]
        if ip not in ip_map:
            ip_map[ip] = {"usernames": set(), "locations": set(), "timestamps": []}
        ip_map[ip]["usernames"].add(login["username"])
        ip_map[ip]["locations"].add(login["location"])
        ip_map[ip]["timestamps"].append(login["timestamp"])

    # Step 2 — check each IP for multiple distinct usernames
    for ip, data in ip_map.items():
        if len(data["usernames"]) >= 2:
            username_list = ", ".join(sorted(data["usernames"]))
            location      = " / ".join(data["locations"])
            severity      = "Critical" if len(data["usernames"]) >= 3 else "High"
            alerts.append({
                "type":        "Multiple Usernames from One IP",
                "severity":    severity,
                "icon":        "🖥️",
                "username":    f"{len(data['usernames'])} accounts",
                "ip":          ip,
                "location":    location,
                "timestamp":   data["timestamps"][0],
                "detail":      f"IP {ip} ({location}) attempted login with "
                               f"{len(data['usernames'])} different usernames: {username_list}.",
                "explanation": "A genuine user only logs in with their own username. When the same IP "
                               "tries multiple different usernames, it usually means an automated "
                               "'credential stuffing' tool is running — trying a list of stolen "
                               "username/password pairs to find one that works."
            })
    return alerts


def detect_success_after_failure():
    """
    Detection 3: Successful Login After Repeated Failures
    A 'Success' record where the same username also has a 'Failed'
    record with 3+ attempts — meaning the attacker eventually
    guessed the correct password.

    Steps:
      1. Build a set of usernames that had 3+ failures.
      2. Find all Success records whose username is in that set.
      3. Raise a Critical alert for each match.
    """
    alerts = []

    # Step 1 — collect all usernames with 3+ failures
    failed_usernames = {
        l["username"]
        for l in login_attempts
        if l["status"] == "Failed" and l["attempts"] >= 3
    }

    # Step 2 — find success records for those usernames
    for login in login_attempts:
        if login["status"] == "Success" and login["username"] in failed_usernames:
            alerts.append({
                "type":        "Successful Login After Repeated Failures",
                "severity":    "Critical",
                "icon":        "🔓",
                "username":    login["username"],
                "ip":          login["ip"],
                "location":    login["location"],
                "timestamp":   login["timestamp"],
                "detail":      f"Account '{login['username']}' had multiple prior failed logins — "
                               f"then a SUCCESSFUL login was recorded at {login['timestamp']} "
                               f"from {login['ip']} ({login['location']}).",
                "explanation": "This pattern means a brute-force or credential stuffing attack SUCCEEDED — "
                               "the attacker guessed the correct password. The account is likely compromised. "
                               "Analysts would immediately disable the account, review what the attacker "
                               "accessed, and determine if data was stolen."
            })
    return alerts


def render_alerts():
    """Collects all 3 detections and prints colour-labelled alert cards."""
    all_alerts = (
        detect_repeated_failures() +
        detect_multi_user_same_ip() +
        detect_success_after_failure()
    )

    print(f"\n{'='*90}")
    print(f"  🚨 ALERTS PANEL  —  {len(all_alerts)} alert(s) detected")
    print(f"{'='*90}")

    if not all_alerts:
        print("  ✅ No alerts detected in this dataset.")
        return

    sev_order = {"Critical": 0, "High": 1, "Medium": 2}
    all_alerts.sort(key=lambda a: sev_order.get(a["severity"], 3))

    for i, alert in enumerate(all_alerts, 1):
        print(f"\n  [{i}] {alert['icon']}  {alert['type']}  [{alert['severity'].upper()}]")
        print(f"       👤 {alert['username']}  |  🌐 {alert['ip']}  |  📍 {alert['location']}  |  🕐 {alert['timestamp']}")
        print(f"       📌 {alert['detail']}")
        print(f"       💡 {alert['explanation']}")
        print(f"  {'─'*85}")

    print(f"{'='*90}\n")

# NOTE: called in Section 10
```

## Section 8 — SHA-256 Hash Demo
Live SHA-256 computation using Python's `hashlib` (equivalent to the browser Web Crypto API).

```python
# ============================================================
# SECTION 8 — SHA-256 HASH DEMO
# Uses Python's built-in hashlib (equivalent to Web Crypto API)
# ============================================================
import hashlib

def compute_hash(text):
    """
    Computes a real SHA-256 hash for the given string.
    Returns the 64-character lowercase hex digest.
    """
    if not text.strip():
        print("[ type something to generate a hash ]")
        return

    # Encode text to bytes, then hash
    hash_hex = hashlib.sha256(text.encode("utf-8")).hexdigest()

    print(f"\n  Input  ({len(text)} character(s)): {repr(text)}")
    print(f"  SHA-256 hash            : {hash_hex}")
    print(f"  Hash length             : {len(hash_hex)} characters (256 bits)")
    print()
    print("  💡 The SHA-256 hash is always 64 characters long, regardless of input length.")
    print("     Try changing just one letter — the entire hash changes completely.")
    print("     This is called the 'avalanche effect' — a key property of secure hash functions.")
    return hash_hex


def render_hash_examples():
    """Prints the pre-loaded fictional file hash table."""
    print(f"\n{'='*110}")
    print("  🔐 HASHED EVIDENCE ITEMS (Pre-computed SHA-256)")
    print(f"{'='*110}")
    header = f"  {'File':<40}  {'SHA-256 Hash':<68}  Notes"
    print(header)
    print(f"  {'─'*105}")
    for ex in hash_examples:
        print(f"  {ex['file']:<40}  {ex['hash']:<68}  {ex['notes']}")
    print(f"{'='*110}\n")


# --- Interactive demo: try different inputs ---
print("=== SHA-256 Live Demo ===")
compute_hash("CyberTrace")
compute_hash("CyberTrace")   # Same input → same hash
compute_hash("cybertrace")   # Different case → completely different hash
render_hash_examples()
```
